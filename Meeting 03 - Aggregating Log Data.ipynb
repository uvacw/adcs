# Meeting 3 - Aggregating Log Data

These steps allow us to aggregate donated data (i.e., at item level) to individual (i.e., respondent) level. Please note that this notebook only contains the commands that were executed during the video tutorial. Please check out the video to get more context. 

```python
import pandas as pd
```

## Loading the relevant datasets

```python
donations = pd.read_excel('dataset_tiktok/tiktok_watch_history.xlsx')
```

```python
site_logs = pd.read_excel('dataset_tiktok/tracking.xlsx')
```

```python
video_metadata = pd.read_csv('class_tiktok/metadata/tiktok_videos.csv')
```

### Creating a common column with video identifier (i.e., video_id)

```python
def return_video_id(link):
    if link:
        return link.split('/')[-2]
    return
```

```python
donations.iloc[0]
```

```python
donations.iloc[0]['Link']
```

```python
# Testing with one example
print(return_video_id(donations.iloc[0]['Link']))
```

```python
donations['video_id'] = donations['Link'].apply(return_video_id)
```

```python
donations.head()
```

```python
donations.dtypes
```

### Merging - videos

```python
dataset = pd.merge(donations, video_metadata, how='left', on='video_id')
```

```python
dataset.head()
```

## Aggregating the dataset at individual level

```python
dataset['participant'].value_counts()
```

### Example 1: Number of videos watched

```python
count_videos_watched = pd.pivot_table(dataset, index='participant', values='Link', aggfunc='count').reset_index()
```

```python
count_videos_watched.head()
```

```python
count_videos_watched
```

```python
count_videos_watched = count_videos_watched.rename(columns={'Link': 'count_videos_watched'})
```

```python
count_videos_watched.head()
```

```python
len(dataset['participant'].unique())
```

```python
len(count_videos_watched)
```

```python
count_videos_watched.to_excel('dataset_tiktok/count_videos_watched.xlsx', index=False)
```

### Example 2: Total video duration 

```python
sum_video_duration = pd.pivot_table(dataset, index='participant', values='duration', aggfunc='sum').reset_index()
```

```python
sum_video_duration.head()
```

```python
sum_video_duration = sum_video_duration.rename(columns={'duration': 'sum_video_duration'})
```

```python
sum_video_duration.head()
```

```python
len(dataset['participant'].unique())
```

```python
len(sum_video_duration)
```

```python
sum_video_duration.to_excel('dataset_tiktok/sum_video_duration.xlsx', index=False)
```

### Example 3: % of 4K videos

```python
dataset['video_quality'].value_counts()
```

```python
dataset['video_quality'].head()
```

```python
def check_4K(video_quality):
    if str(video_quality) == '4k':
        return 1 
    return 0
```

```python
dataset['is_4K'] = dataset['video_quality'].apply(check_4K)
```

```python
dataset['is_4K'].value_counts()
```

```python
perc_4K_videos = pd.pivot_table(dataset, index='participant', values='is_4K', aggfunc='mean').reset_index()
```

```python
perc_4K_videos.head()
```

```python
perc_4K_videos = perc_4K_videos.rename(columns={'is_4K': 'perc_4K_videos'})
```

```python
perc_4K_videos.head()
```

```python
perc_4K_videos.to_excel('dataset_tiktok/perc_4K_videos.xlsx', index=False)
```

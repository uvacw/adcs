## Statistical analysis

Note: this is a general example of analysis. The usage of these variables may not make full sense.
```python
X = analysis[['personality', 'leeftijd', 'geslacht']]
X = sm.add_constant(X)
y = analysis['count_videos_watched']
model = sm.OLS(y, X).fit()
```

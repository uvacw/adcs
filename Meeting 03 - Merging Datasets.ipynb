```python
import pandas as pd
```

## Loading datasets

```python
survey = pd.read_excel('class_tiktok/ADCS-2026-class-tiktok_September+14,+2026_11.27.xlsx', skiprows=[1,])
```

```python
perc_4K_videos = pd.read_excel('dataset_tiktok/perc_4K_videos.xlsx')
```

```python
sum_video_duration = pd.read_excel('dataset_tiktok/sum_video_duration.xlsx')
```

```python
count_videos_watched = pd.read_excel('dataset_tiktok/count_videos_watched.xlsx')
```

```python
site_logs_unique = pd.read_excel('dataset_tiktok/site_logs_unique.xlsx')
```

## Merging

```python
dataset = survey.merge(perc_4K_videos, how='left', on='participant')
```

```python
dataset.head()
```

```python
dataset = pd.merge(dataset, sum_video_duration, how='left', on='participant')
```

```python
dataset.head()
```

```python
dataset = dataset.merge(count_videos_watched, how='left', on='participant')
```

```python
dataset.head()
```

```python
dataset = dataset.merge(site_logs_unique, how='left', on='participant')
```

```python
dataset.head()
```
